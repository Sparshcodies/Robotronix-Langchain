```python
import matplotlib.pyplot as plt

# Data
models = ["YOLO26n", "YOLO26s", "YOLO26m", "YOLO26l", "YOLO26x"]
map_val = [40.9, 48.6, 53.1, 55.0, 57.5]
flops = [5.4, 20.7, 68.2, 86.4, 193.9]
inference_t4 = [1.7, 2.5, 4.7, 6.2, 11.8]

# Scatter plot: Accuracy vs FLOPs (marker size = inference time)
plt.figure()
plt.scatter(flops, map_val, s=[t * 40 for t in inference_t4])

for i, model in enumerate(models):
    plt.annotate(model, (flops[i], map_val[i]), textcoords="offset points", xytext=(5,5))

plt.xlabel("FLOPs (B)")
plt.ylabel("mAP@50-95")
plt.title("Accuracy vs FLOPs (Marker Size = Inference Time on T4 ms)")

plt.show()
```
